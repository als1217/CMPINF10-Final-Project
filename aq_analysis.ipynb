```python
import pandas as pd

air_quality_analysis = pd.read_csv("38e85e2d-26a9-4ae1-bcb9-d9feaceeab40.csv", sep="\t")

# Turns air_quality_analysis, which only has one column, into a usable DataFrame with a column for each data point
dataframe_list = []
for i in range(10800):
    row_list = air_quality_analysis.iat[i, 0].split(",")
    dataframe_list.append(row_list)
df = pd.DataFrame(dataframe_list, index=range(10800), columns=["_id", "datetime_utc", "datetime_et", "latitude", "longitude", "parameter", "value", "units", "aqi", "aqi_descriptor", "aqi_type", "place",
                                                               "latitude_str", "longitude_str", "date_of_prediction"])
# Compiles a list of every neighborhood in Pittsburgh for which there is data
neighborhood_list = []
for i in range(10800):
    if df.iat[i, 11] not in neighborhood_list:
        neighborhood_list.append(df.iat[i, 11])
# Removes an empty string from the list of neighborhoods. We're not going to work with data that doesn't correspond to a particular neighborhood
neighborhood_list.remove('""')

# Creates a list of parameters(types of air pollutants) for which there is data
parameter_list = ["CO", "NO2", "PM25_RH35"]
# Creates lists of data points and corresponding units for observations. These will be filled with data from each neighborhood, printed and cleared.
neighborhood_CO_data_list = []
neighborhood_CO_unit_list = []
neighborhood_NO2_data_list = []
neighborhood_NO2_unit_list = []
neighborhood_PM25_RH35_data_list = []
neighborhood_PM25_RH35_unit_list = []
for neighborhood in neighborhood_list:
    for i in range(10800):
        if df.iat[i, 11] == neighborhood:
            if df.iat[i, 5] == "CO":
                neighborhood_CO_data_list.append(df.iat[i, 6])
                if df.iat[i, 7] not in neighborhood_CO_unit_list:
                    neighborhood_CO_unit_list.append(df.iat[i, 7])
            elif df.iat[i, 5] == "NO2":
                neighborhood_NO2_data_list.append(df.iat[i, 6])
                if df.iat[i, 7] not in neighborhood_NO2_unit_list:
                    neighborhood_NO2_unit_list.append(df.iat[i, 7])
            else:
                neighborhood_PM25_RH35_data_list.append(df.iat[i, 6])
                if df.iat[i, 7] not in neighborhood_PM25_RH35_unit_list:
                    neighborhood_PM25_RH35_unit_list.append(df.iat[i, 7])

# Each of these if/else statements checks if the collections of observations for each neighborhood/parameter combinationare in the same unit. If they're not, they tell the user that some units need to be 
# converted in order for an average value to make sense. Fortunately for me, all of the averages make sense
    if len(neighborhood_CO_unit_list) == 1:
        total = 0
        count = len(neighborhood_CO_data_list)
        for i in range(len(neighborhood_CO_data_list)):
            total += float(neighborhood_CO_data_list[i])
        print("Average prediction for CO in " + neighborhood + ": " + str(total/count) + " " + str(neighborhood_CO_unit_list))
    else:
        print("Prediction for CO in " + neighborhood + " needs unit conversion")

    if len(neighborhood_NO2_unit_list) == 1:
        total = 0
        count = len(neighborhood_NO2_data_list)
        for i in range(len(neighborhood_NO2_data_list)):
            total += float(neighborhood_NO2_data_list[i])
        print("Average prediction for NO2 in " + neighborhood + ": " + str(total/count) + " " + str(neighborhood_NO2_unit_list))
    else:
        print("Prediction for NO2 in " + neighborhood + " needs unit conversion")

    if len(neighborhood_PM25_RH35_unit_list) == 1:
        total = 0
        count = len(neighborhood_PM25_RH35_data_list)
        for i in range(len(neighborhood_PM25_RH35_data_list)):
            total += float(neighborhood_PM25_RH35_data_list[i])
        print("Average prediction for PM25_RH35 in " + neighborhood + ": " + str(total/count) + " " + str(neighborhood_PM25_RH35_unit_list))
    else:
        print("Prediction for PM25_RH35 in " + neighborhood + " needs unit conversion")
    """
    print(neighborhood + " CO data: " + str(neighborhood_CO_data_list))
    print(neighborhood + " NO2 data: " + str(neighborhood_NO2_data_list))
    print(neighborhood + " PM25-RH35 data: " + str(neighborhood_PM25_RH35_data_list))
    """
    
    neighborhood_CO_data_list.clear()
    neighborhood_NO2_data_list.clear()
    neighborhood_PM25_RH35_data_list.clear()
```

```python
import pandas as pd
import statistics

# Creates a usable DataFrame of the data
air_quality_analysis = pd.read_csv("38e85e2d-26a9-4ae1-bcb9-d9feaceeab40.csv")
df = pd.DataFrame(air_quality_analysis)
# print(air_quality_analysis)
# print(df)

# Creates a list of every Pittsburgh neighborhood represented in the data
neighborhood_list = []
for i in range(10800):
    if df.iat[i, 11] not in neighborhood_list:
        neighborhood_list.append(df.iat[i, 11])
# print(neighborhood_list)

# Checks that the only parameters are CO, NO2 and PM25_RH35(small particulate matter). We do not want any NaN values in parameter/pollutant!
parameter_list = []
for i in range(10800):
    if df.iat[i, 5] not in parameter_list:
        parameter_list.append(df.iat[i, 5])
# print(parameter_list)

# The units for measurements of each pollutant. Will be used throughout the code
CO_unit = "ppm"
NO2_unit = "ppb"
PM25_RH35_unit = "ug m-3"

# Creates lists for the average concentration of each pollutant in each county. These will be iterated through to determine which county has the lowest concentration of each pollutant, and therefore the 
# best air quality based on the data from this dataset.
CO_data_averages = []
NO2_data_averages = []
PM25_RH35_data_averages = []

# Lists of the counties with data for each pollutant. Created to avoid calculating means for empty lists.
neighborhoods_with_CO_data = []
neighborhoods_with_NO2_data = []
neighborhoods_with_PM25_RH35_data = []

# Iterates through the list of neighborhoods and compiles lists of data points and units for each neighborhood. At the end of each loop, it clears the lists so they can be used again for the next
# neighborhood. During the loops, it adds to a complete list of average concentrations of each pollutant for each neighborhood.
for neighborhood in neighborhood_list:
    # Creates lists for data points and unit information for each pollutant. These will each be filled up once for each neighborhood.
    neighborhood_CO_data_list = []
    neighborhood_NO2_data_list = []
    neighborhood_PM25_RH35_data_list = []
    for i in range(10800):
        if df.iat[i, 11] == neighborhood:
            if df.iat[i, 5] == "CO":
                neighborhood_CO_data_list.append(df.iat[i, 6])
            elif df.iat[i, 5] == "NO2":
                neighborhood_NO2_data_list.append(df.iat[i, 6])
            else:
                neighborhood_PM25_RH35_data_list.append(df.iat[i, 6])
        # print(neighborhood + " CO data: ")
        # print(str(neighborhood_CO_data_list))
        # print(neighborhood + " NO2 data: ")
        # print(str(neighborhood_NO2_data_list))
        # print(neighborhood + " PM25_RH35 data: ")
        # print(str(neighborhood_PM25_RH35_data_list))
        if len(neighborhood_CO_data_list) > 0:
            CO_average = statistics.mean(neighborhood_CO_data_list)
            CO_data_averages.append(CO_average)
            neighborhoods_with_CO_data.append(neighborhood)
        if len(neighborhood_NO2_data_list) > 0:
            NO2_average = statistics.mean(neighborhood_NO2_data_list)
            NO2_data_averages.append(NO2_average)
            neighborhoods_with_NO2_data.append(neighborhood)
        if len(neighborhood_PM25_RH35_data_list) > 0:
            PM25_RH35_average = statistics.mean(neighborhood_PM25_RH35_data_list)
            PM25_RH35_data_averages.append(PM25_RH35_average)
            neighborhoods_with_PM25_RH35_data.append(neighborhood)
        # print("Average CO concentration for " + neighborhood + " in " + CO_unit + ": " + str(CO_average))
        # print("Average NO2 concentration for " + neighborhood + " in " + NO2_unit + ": " + str(NO2_average))
        # print("Average PM25-RH35 concentration for " + neighborhood + " in " + PM25_RH35_unit + ": " + str(PM25_RH35_average))
        """
        print(neighborhood_CO_data_list)
        print(neighborhood_NO2_data_list)
        print(neighborhood_PM25_RH35_data_list)
        """

# Prints the neighborhoods with the lowest concentration of each pollutant. The index_of variables correspond to the index at which the lowest concentration was found in the list of averages, 
# and are used to more efficiently fetch the corresponding neighborhood.
lowest_avg_CO_concentration = min(CO_data_averages)
CO_index_of = CO_data_averages.index(lowest_avg_CO_concentration)
lowest_avg_NO2_concentration = min(NO2_data_averages)
NO2_index_of = NO2_data_averages.index(lowest_avg_NO2_concentration)
lowest_avg_PM25_RH35_concentration = min(PM25_RH35_data_averages)
PM25_RH35_index_of = PM25_RH35_data_averages.index(lowest_avg_PM25_RH35_concentration)

print("Neighborhood with the lowest average concentration of CO: " + neighborhoods_with_CO_data[CO_index_of] + " with a concentration of " + str(lowest_avg_CO_concentration) + " " + CO_unit)
print("Neighborhood with the lowest average concentration of NO2: " + neighborhoods_with_NO2_data[NO2_index_of] + " with a concentration of " + str(lowest_avg_NO2_concentration) + " " + NO2_unit)
print("Neighborhood with the lowest average concentration of PM25-RH35: " + neighborhoods_with_PM25_RH35_data[PM25_RH35_index_of] + " with a concentration of " + str(lowest_avg_PM25_RH35_concentration) + " " + PM25_RH35_unit)

# print(neighborhood_list)
# print(CO_data_averages)
# print(NO2_data_averages)
# print(PM25_RH35_data_averages)

# West Finley has the lowest CO and NO2 concentration, but Ohioville has the lowest small particulate matter(PM25-RH35) concentration. I decided to print out the rest of Ohioville and West Finley's data
# to get the full picture:

print("West Finley average concentration of PM25-RH35: " + str(PM25_RH35_data_averages[neighborhoods_with_PM25_RH35_data.index("West Finley")]) + " " + PM25_RH35_unit)
print("Ohioville average concentration of CO: " + str(CO_data_averages[neighborhoods_with_CO_data.index("Ohioville")]) + " " + CO_unit)
print("Ohioville average concentration of NO2: " + str(NO2_data_averages[neighborhoods_with_NO2_data.index("Ohioville")]) + " " + NO2_unit)

# The difference between Ohioville's and West Finley's average concentration of CO is only a few thousandths of a ppm, which is very little. In addition, its average concentration of NO2 is about 25% more 
# than that of West Finley. However, West Finley's average concentration of small particulate matter is approximately three times more than that of Ohioville. From the relative air quality between the two
# neighborhoods with the cleanest air, I conclude that Ohioville is the best neighborhood in Pittsburgh in terms of air quality.
```
